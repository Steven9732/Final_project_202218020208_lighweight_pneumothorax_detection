```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Internal test set data
# -----------------------------
metrics = ['Accuracy', 'Precision', 'Recall', 'F1-score', 'PR-AUC']

data = {
    'Baseline': [91.37, 83.54, 76.03, 79.61, 87.46],
    'Baseline + ECA': [91.04, 76.41, 86.14, 80.99, 88.45],
    'Baseline + LSMF': [92.03, 85.77, 76.78, 81.03, 89.72],
    'Baseline + ECA + LSMF': [91.54, 79.15, 83.90, 81.45, 89.08],
}

models = list(data.keys())
values = np.array(list(data.values()))  # shape = (4, 5)

# -----------------------------
# Plot settings
# -----------------------------
x = np.arange(len(metrics))
width = 0.18

plt.figure(figsize=(10, 6), dpi=300)

for i, model in enumerate(models):
    plt.bar(
        x + (i - 1.5) * width,
        values[i],
        width=width,
        label=model,
        edgecolor='black',
        linewidth=0.8
    )

# -----------------------------
# Add value labels
# -----------------------------
for i, model in enumerate(models):
    for j, v in enumerate(values[i]):
        plt.text(
            x[j] + (i - 1.5) * width,
            v + 0.25,
            f'{v:.2f}',
            ha='center',
            va='bottom',
            fontsize=8,
            rotation=90
        )

# -----------------------------
# Axis / legend / style
# -----------------------------
plt.xticks(x, metrics, fontsize=11)
plt.yticks(fontsize=10)
plt.ylabel('Performance (%)', fontsize=12)
plt.xlabel('Metrics', fontsize=12)
plt.title('Ablation Study on Internal Test Set', fontsize=13)
plt.ylim(0, 100)

plt.legend(
    fontsize=9,
    frameon=True,
    loc='upper center',
    bbox_to_anchor=(0.5, 1.18),
    ncol=2
)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()

# save figure
plt.savefig('internal_ablation_grouped_bar.png', dpi=600, bbox_inches='tight')
plt.savefig('internal_ablation_grouped_bar.pdf', bbox_inches='tight')

plt.show()
```
